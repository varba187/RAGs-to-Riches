```python
import pandas as pd

bart_df = pd.read_csv("bart_baseline_train_eval_results.csv")
rag_sequence_df = pd.read_csv("rag_sequence_results.csv")
rag_token_df = pd.read_csv("rag_token_results.csv")

bart_accuracy = bart_df["em"].mean() * 100
rag_sequence_accuracy = rag_sequence_df["em"].mean() * 100
rag_token_accuracy = rag_token_df["em"].mean() * 100

comparison_df = pd.DataFrame([
    {"model": "BART baseline", "EM": bart_accuracy},
    {"model": "RAG-Sequence", "EM": rag_sequence_accuracy},
    {"model": "RAG-Token", "EM": rag_token_accuracy},
])

comparison_df.to_csv("final_comparison.csv", index = False)
comparison_df
```
